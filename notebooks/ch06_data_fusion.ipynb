## ۶-۲) همان NDVI، این‌بار با کد


```python
import rasterio
import numpy as np

with rasterio.open("B04.jp2") as red_src:          # red band
    red = red_src.read(1).astype("float32")
    profile = red_src.profile                      # keep CRS, size, ...

with rasterio.open("B08.jp2") as nir_src:          # near-infrared
    nir = nir_src.read(1).astype("float32")

# compute NDVI for the WHOLE image at once (no loops!)
denominator = nir + red
ndvi = np.where(denominator == 0, 0, (nir - red) / denominator)

profile.update(dtype="float32", count=1)
with rasterio.open("ndvi.tif", "w", **profile) as dst:
    dst.write(ndvi, 1)

print("NDVI min:", f"{float(ndvi.min()):.2f}",
      "| max:", f"{float(ndvi.max()):.2f}")
```
