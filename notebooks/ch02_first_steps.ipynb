## ۲-۲) اولین نقشه‌ی شما در کد


```python
!pip install geopandas -q

import geopandas as gpd

# GADM publishes free administrative boundaries for every country.
# level 1 = provinces. The file is read straight from the web.
url = "https://geodata.ucdavis.edu/gadm/gadm4.1/json/gadm41_IRN_1.json"
provinces = gpd.read_file(url)

print(type(provinces))       # a GeoDataFrame: a table with a geometry column
print(provinces.shape)       # (31, ...) -> 31 provinces
provinces.head(3)
```

```python
provinces.plot(figsize=(8, 8), edgecolor="white", linewidth=0.6)

# and one province on its own:
tehran = provinces[provinces["NAME_1"] == "Tehran"]
tehran.plot(figsize=(4, 4), color="#2E75B6")
```

## ۲-۴) چاپ روی صفحه: تابع print و f-string


```python
city_name = "Tehran"
latitude = 35.6892

# 1) concatenation: works, but note str() and the manual spaces
print("The city " + city_name + " is at latitude " + str(latitude) + ".")
```

```python
# 2) format(): no str() needed, but the values sit far from their slots
print("The city {} is at latitude {}.".format(city_name, latitude))
```

```python
print("Welcome to GeoAI Engineering!")
print(latitude)
# 3) f-string: the value sits exactly where you read it
print(f"The city {city_name} is at latitude {latitude}.")
```

## ۲-۱۱) مطالعه‌ی آزاد (اختیاری): اولین طعم یک شاخص ماهواره‌ای


```python
nir_band = 0.45   # near-infrared reflectance of one pixel
red_band = 0.12   # red reflectance of the same pixel
ndvi = (nir_band - red_band) / (nir_band + red_band)
print(f"{ndvi:.3f}")   # 0.579  -- format for display, do not round the value
```
