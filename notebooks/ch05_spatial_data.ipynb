## خودهمبستگی مکانی (Spatial Autocorrelation)


```python

!pip install esda libpysal

import geopandas as gpd
from libpysal.weights import KNN
from esda.moran import Moran

gdf = gpd.read_file("air_quality_stations.gpkg").to_crs(epsg=32639)  # metric

# 1) who counts as a "neighbour"? here: the 8 nearest stations
w = KNN.from_dataframe(gdf, k=8)
w.transform = "r"        # row-standardise: every row of weights sums to 1

# 2) Moran's I of the pollution column, tested against 999 random shuffles
mi = Moran(gdf["pm25"].values, w, permutations=999)

# ~1 clustered, ~0 random, <0 chequerboard
print(f"Moran's I = {mi.I:.3f}")
print(f"p-value   = {mi.p_sim:.4f}")  # < 0.05 -> the pattern is NOT random
```
