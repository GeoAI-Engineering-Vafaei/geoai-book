## ۳-۱۴) پروژه‌ی پایان فصل: تحلیل ساده‌ی دسترسی به بیمارستان‌ها


```python
crisis_point = {"lat": 35.7200, "lon": 51.3800}
hospitals = [
    {"name": "Milad", "lat": 35.7350, "lon": 51.3650, "capacity": 1000},
    {"name": "Shariati", "lat": 35.7400, "lon": 51.4200, "capacity": 450},
    {"name": "Local", "lat": 35.7800, "lon": 51.4500, "capacity": 50},
    {"name": "Heart", "lat": 35.7100, "lon": 51.4000, "capacity": 350},
    {"name": "Imam", "lat": 35.7050, "lon": 51.3900, "capacity": 600},
]
```

```python
from math import radians, cos

def approx_distance(lat1, lon1, lat2, lon2):
    """A simple km approximation, good only for short distances."""
    # the middle latitude of the pair:
    mean_lat = radians((lat1 + lat2) / 2)
    d_lat = (lat1 - lat2) * 111                # 1 degree of latitude ~ 111 km
    d_lon = (lon1 - lon2) * 111 * cos(mean_lat)  # meridians converge poleward
    return (d_lat ** 2 + d_lon ** 2) ** 0.5
```

```python
results = []
for h in hospitals:
    try:
        dist = approx_distance(crisis_point["lat"], crisis_point["lon"],
                               h["lat"], h["lon"])
        if dist > 10:
            continue
        if h["capacity"] < 100:
            continue
        results.append({"name": h["name"], "distance": dist,
                        "capacity": h["capacity"]})
    except (KeyError, TypeError) as err:
        print(f"Invalid record skipped: {err}")
        continue
results.sort(key=lambda r: r["distance"])
```

```python
print(f"Report: {len(results)} suitable hospital(s) found")
for i, r in enumerate(results, start=1):
    print(f"{i}. {r['name']:10s} | {r['distance']:.2f} km "
          f"| {r['capacity']} beds")
if len(results) == 0:
    print("No suitable hospital found.")
```
